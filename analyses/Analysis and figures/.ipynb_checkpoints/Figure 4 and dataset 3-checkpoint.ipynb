A function that goes through each of the strains file, extracts COGs, CK ID, KEGGs and its description, then concats all of the datafames

```python
strains_lst=['MED4', 'MIT9515', 'MIT9202', 'MIT9215', 'MIT0604', 'MIT9301', 'SB', 'AS9601', 'MIT9312',
             'NATL1A', 'NATL2A', 'PAC1', 'MIT9313']


def extract_cog_numbers(x):
    cogs = re.findall(r'COG(\d+)', x)
    cog_numbers = ['COG' + number for number in cogs]  # Convert to COG####
    return cog_numbers

def extract_kegg_numbers(x):
    keggs = re.findall(r'kegg=([^;]+)', x)
    return keggs   

def extract_kegg_description(x):
    keggs_descrip = re.findall(r'kegg_description=([^;]+)', x)
    return keggs_descrip

def extract_CK_ID(x):
    cogs = re.findall(r'CK_(\d+)', x)
    cog_numbers = ['CK_' + number for number in cogs]  
    return cog_numbers



def extract_eggNOG_description(x):
    eggNOG_description = re.findall(r'COG: ([^,]+)', x)
    return ', '.join(eggNOG_description)  

def open_file(strains_lst):
    dfs = {}  # Dictionary to store dataframes for each strain
    for i in strains_lst:
        gff_file_path = r'C:\Users\yaras\OneDrive\Desktop\Analysis for paper\Python Code\gffs\Pro_{}.gff'.format(i)
        column_names = ['seqid', 'source', 'type', 'start', 'end', 'score', 'strand', 'phase', 'attributes']
        df = pd.read_csv(gff_file_path, sep='\t', comment='#', names=column_names)
        # Apply functions to extract additional information
        df['cog_numbers'] = df['attributes'].apply(extract_cog_numbers)
        df['CK_ID'] = df['attributes'].apply(extract_CK_ID)
        df['kegg_numbers'] = df['attributes'].apply(extract_kegg_numbers)
        df['eggNOG_description'] = df['attributes'].apply(extract_eggNOG_description)
        df['kegg_description'] = df['attributes'].apply(extract_kegg_description)
        df['strain']=i
        df=df.explode('cog_numbers', ignore_index=False)
        # Save dataframe in the dictionary with the strain name as the key
        dfs[i] = df
    return dfs


dfs = open_file(strains_lst)


strains_df = pd.concat(dfs.values(), ignore_index=True)


strains_df
```

```python
""" here we will take each pair of strains and check each COG within this pair, i use the pivot_df dataframe because the 
index are the COGs and the coulmns are the strain so it is stored exactly how i want it """

def mann_whitney_test(df):

    
    # Get all strain names and COGs
    n_sen=['MED4', 'MIT9312', 'MIT0604','MIT9515','MIT9215','AS9601','MIT9202','NATL1A']
    other_sen= ['SB', 'PAC1', 'NATL2A', 'MIT1327', 'MIT9313', 'MIT9301']
    cogs = df.index

    # Create an empty list to store results
    results = []

    # Iterate through all pairs of strains for each COG

    for cog in cogs:
        # Get the COG values for the two strains
        data1 = df.loc[cog, n_sen]
        data2 = df.loc[cog, other_sen]


        # Perform the Mann-Whitney U test
        stat, p_value = mannwhitneyu(data1, data2)


        
        # Store the results
        results.append([cog, stat, p_value])
        
        
    # Convert the results to a dataframe
    results_df = pd.DataFrame(results, columns=['cog_numbers', 'U_statistic', 'p_value'])
    
    # Calculation adjusted p_values
#     adjusted_p_values=fdrcorrection(results_df['p_value'])[1]
    adjusted_p_values = multipletests(results_df['p_value'], method='fdr_bh')[1]
    results_df['adjusted_p_values']=adjusted_p_values

    return results_df


# Perform the Mann-Whitney U test and get the results dataframe
results_df = mann_whitney_test(pivot_df)

results_df

```

```python

def mann_whitney_test(df):
    # Define the strain groups
    n_sen = ['MED4', 'MIT9312', 'MIT0604', 'MIT9515', 'MIT9215', 'AS9601', 'MIT9202', 'NATL1A']
    other_sen = ['SB', 'PAC1', 'NATL2A', 'MIT1327', 'MIT9313', 'MIT9301']
    cogs = df.index

    # Create an empty list to store results
    results = []

    # Iterate through all COGs
    for cog in cogs:
        # Get the COG values for the two strain groups
        data1 = df.loc[cog, n_sen]
        data2 = df.loc[cog, other_sen]

        # Perform the Mann-Whitney U test
        stat, p_value = mannwhitneyu(data1, data2)

        # Determine the direction of the effect
        median_n = data1.median()
        median_other = data2.median()
        if median_n > median_other:
            direction = 'N'
        else:
            direction = 'Other'

        # Store the results
        results.append([cog, stat, p_value, median_n, median_other, direction])

    # Convert the results to a DataFrame
    results_df = pd.DataFrame(results, columns=['cog_numbers', 'U_statistic', 'p_value', 'median_N', 'median_Other', 'Direction'])

    # Adjust p-values for multiple testing
    adjusted_p_values = multipletests(results_df['p_value'], method='fdr_bh')[1]
    results_df['adjusted_p_values'] = adjusted_p_values

    return results_df

# Perform the Mann-Whitney U test and get the results dataframe
results_df = mann_whitney_test(pivot_df)

results_df
```

```python

def mann_whitney_test(df):
    # Define the strain groups
    HL = ['MED4', 'MIT9312', 'MIT0604', 'MIT9515', 'MIT9215', 'AS9601', 'MIT9202', 'SB', 'MIT9301']
    LL = ['NATL1A','PAC1', 'NATL2A', 'MIT1327', 'MIT9313']
    cogs = df.index

    # Create an empty list to store results
    results = []

    # Iterate through all COGs
    for cog in cogs:
        # Get the COG values for the two strain groups
        data1 = df.loc[cog, HL]
        data2 = df.loc[cog, LL]

        # Perform the Mann-Whitney U test
        stat, p_value = mannwhitneyu(data1, data2)

        # Determine the direction of the effect
        median_HL = data1.median()
        median_LL = data2.median()
        if median_HL > median_LL:
            direction = 'HL'
        else:
            direction = 'LL'

        # Store the results
        results.append([cog, stat, p_value, median_HL, median_LL, direction])

    # Convert the results to a DataFrame
    results_df_ecotypes = pd.DataFrame(results, columns=['cog_numbers', 'U_statistic', 'p_value', 'median_HL', 'median_LL', 'Direction'])

    # Adjust p-values for multiple testing
    adjusted_p_values = multipletests(results_df['p_value'], method='fdr_bh')[1]
    results_df_ecotypes['adjusted_p_values'] = adjusted_p_values

    return results_df_ecotypes

# Perform the Mann-Whitney U test and get the results dataframe
results_df_ecotypes = mann_whitney_test(pivot_df)

results_df_ecotypes
```

```python

# Create the contingency table for each function
contingency_table = pd.crosstab(combined_dfs_cogs['eggNOG_description'], 
                                combined_dfs_cogs['Category'], 
                                values=combined_dfs_cogs['count_over_total'], 
                                aggfunc='sum', 
                                dropna=False)

# Initialize a dictionary to store results for each function and its pairwise tests
pairwise_results = {}

# Loop through each function (eggNOG_description)
for func in contingency_table.index:
    # Extract the counts for the function
    table = contingency_table.loc[func]
    
    # Drop NaN categories from the table
    table = table.dropna()

    # Generate all pairwise category combinations
    category_pairs = itertools.combinations(table.index, 2)
    
    # Initialize a dictionary to store p-values for each category pair for the current function
    p_values_for_func = {}

    # Perform Fisher's Exact Test for each pair
    for cat1, cat2 in category_pairs:
        # Create the 2x2 contingency table for this category pair
        # Use the counts of the two categories being compared
        contingency = [[table[cat1], table[cat2]], 
                       [sum(table) - table[cat1], sum(table) - table[cat2]]]
        
        # Apply Fisher's Exact Test
        _, p_value = fisher_exact(contingency)
        
        # Store the p-value for the category pair
        p_values_for_func[f"{cat1} vs {cat2}"] = p_value

    # Store results for the current function
    pairwise_results[func] = p_values_for_func

# Convert the pairwise results into a DataFrame for better visualization
pairwise_results_df = pd.DataFrame(pairwise_results)

# Display the p-values for each function and its pairwise category comparisons
pairwise_results_df
```
